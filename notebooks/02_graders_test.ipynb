```python
# 1. Imports
import sys
import os
from importlib import reload
from dotenv import load_dotenv

_PROJECT_ROOT = os.path.abspath('..')
if _PROJECT_ROOT not in sys.path:
    sys.path.insert(0, _PROJECT_ROOT)
load_dotenv(os.path.join(_PROJECT_ROOT, '.env'))

from src import graders
reload(graders)
from src.graders import retrieval_grader, hallucination_grader, answer_grader, question_rewriter

print("--- TESTING GRADERS --- \n")

# ---------------------------------------------------
# Test 1: Document Grader
# ---------------------------------------------------
print("1. Testing Document Grader...")
question = "What is Self-RAG?"
doc_relevant = "Self-RAG is a new framework to improve LLM generation quality through retrieval and self-reflection."
doc_irrelevant = "Bananas are a great source of potassium and are grown in tropical climates."

score_yes = retrieval_grader.invoke({"question": question, "document": doc_relevant})
score_no = retrieval_grader.invoke({"question": question, "document": doc_irrelevant})

print(f"  Relevant Doc Score (Should be 'yes'): {score_yes.binary_score}")
print(f"  Irrelevant Doc Score (Should be 'no'): {score_no.binary_score}\n")

# ---------------------------------------------------
# Test 2: Hallucination Grader
# ---------------------------------------------------
print("2. Testing Hallucination Grader...")
facts = "The company revenue in Q3 was $50 million."
generation_grounded = "In the third quarter, the company made 50 million dollars."
generation_hallucinated = "The company revenue in Q3 was $50 million, and they acquired Apple for $1 billion."

score_grounded = hallucination_grader.invoke({"documents": facts, "generation": generation_grounded})
score_hallucinated = hallucination_grader.invoke({"documents": facts, "generation": generation_hallucinated})

print(f"  Grounded Answer Score (Should be 'yes'): {score_grounded.binary_score}")
print(f"  Hallucinated Answer Score (Should be 'no'): {score_hallucinated.binary_score}\n")

# ---------------------------------------------------
# Test 3: Answer Grader
# ---------------------------------------------------
print("3. Testing Answer Grader...")
question = "How do I reset my password?"
good_answer = "To reset your password, click on the 'Forgot Password' link on the login page."
bad_answer = "Passwords are a method of security used to protect accounts."

score_resolved = answer_grader.invoke({"question": question, "generation": good_answer})
score_unresolved = answer_grader.invoke({"question": question, "generation": bad_answer})

print(f"  Good Answer Score (Should be 'yes'): {score_resolved.binary_score}")
print(f"  Bad Answer Score (Should be 'no'): {score_unresolved.binary_score}\n")

# ---------------------------------------------------
# Test 4: Query Rewriter
# ---------------------------------------------------
print("4. Testing Query Rewriter...")
bad_query = "ahhh idk what is self rag like how it work"
better_query = question_rewriter.invoke({"question": bad_query})

print(f"  Original Query: {bad_query}")
print(f"  Rewritten Query: {better_query.query}\n")

print("--- TESTING COMPLETE ---")
```
