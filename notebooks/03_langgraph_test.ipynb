```python
import sys
import os
from dotenv import load_dotenv

_PROJECT_ROOT = os.path.abspath('..')
if _PROJECT_ROOT not in sys.path:
    sys.path.insert(0, _PROJECT_ROOT)
load_dotenv(os.path.join(_PROJECT_ROOT, '.env'))

from src.graph import app


def run_test(title: str, question: str):
    print("=" * 50)
    print(title)
    print("=" * 50)
    inputs = {"question": question, "retries": 0, "web_search_count": 0}
    final_state = dict(inputs)

    for output in app.stream(inputs):
        for node, update in output.items():
            final_state.update(update)

    print("\nFINAL ANSWER:")
    print(final_state.get("generation", "(no generation produced)"))
    print(f"Retries: {final_state.get('retries', 0)} | Web searches: {final_state.get('web_search_count', 0)}")
    print()


run_test(
    "TEST 1: In-Domain Question (Should use Vector DB)",
    "What is the concept of Self-RAG?",
)

run_test(
    "TEST 2: Out-of-Domain Question (Should REJECT docs and use WEB SEARCH)",
    "What is the weather in Tokyo today?",
)
```
