```python
import sys
import os
from dotenv import load_dotenv

_PROJECT_ROOT = os.path.abspath("..")
if _PROJECT_ROOT not in sys.path:
    sys.path.insert(0, _PROJECT_ROOT)
load_dotenv(os.path.join(_PROJECT_ROOT, ".env"))

from src.cache import clear_cache
from src.main import process_query

# Start from a clean cache so Test 1 is always a miss
clear_cache()

query_1 = "What is the primary purpose of Self-RAG?"
query_2 = "What's the main goal of the Self-RAG framework?"

print("\n>>> TEST 1: First ask (expect CACHE MISS + slow agent)")
result_1 = process_query(query_1)
assert not result_1["from_cache"], "Test 1 should be a cache miss"
assert result_1["latency_seconds"] > 0.5, "Test 1 should run the full agent"
print("✅ Test 1 passed")

print("\n>>> TEST 2: Exact repeat (expect CACHE HIT + fast)")
result_2 = process_query(query_1)
assert result_2["from_cache"], "Test 2 should be a cache hit"
assert result_2["latency_seconds"] < 0.5, "Test 2 should be much faster than the agent"
assert result_2["answer"] == result_1["answer"], "Test 2 should return the same cached answer"
print("✅ Test 2 passed")

print("\n>>> TEST 3: Rephrased similar question (expect CACHE HIT + fast)")
result_3 = process_query(query_2)
assert result_3["from_cache"], "Test 3 should be a semantic cache hit"
assert result_3["latency_seconds"] < 0.5, "Test 3 should be much faster than the agent"
print("✅ Test 3 passed")

print("\n--- ALL CACHE TESTS PASSED ---")
```
